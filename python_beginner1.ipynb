### 格式化输出

### 不同类型数据使用不同的格式化字符

%c  字符

%s  字符串

%d  十进制整数

%f  浮点数

```python
# 输出为 My name is 007, my age is 23

# %格式化输出
print("My name is %s, my age is %d" % ("007", 23)) 

# format()格式化输出
print("My name is {}, my age is {}".format("007", 23))

# f表达式输出
name, age = "007", 23
print(f"my name is {name}, my age is {age}")
```

### 4. Dict 字典

Key-Value Pair（键值对）保存数据。  

例如：  
student = {  
    "name": "Alice",  
    "age": 22,  
    "major": "Computer Science"  
}  

结构：
Key      Value  

name  → Alice  
age   → 22  
major → Computer Science  

##### 4.1 Dict 查

student = {
    "name": "Alice",
    "age": 22
}  
print(student["name"])  # Alice

get()  
print(student.get("name"))  
如果 key 不存在，则会返回None,而不是KeyError:  
print(student.get("grade"))

keys()  
查看所有 key：  
print(student.keys()) # dict_keys(['name', 'age'])

values()  
查看所有 value：  
print(student.values())  # dict_values(['Alice', 22])

items()  
查看所有 key-value：  
print(student.items())  
#items常用于循环，例如：  
for key, value in student.items():  
    print(key, value)  





```python
student = {
    "name": "Alice",
    "age": 22
}

print(student.get("name"))
print(student.keys())
print(student.values())
print(student.items())
for key, value in student.items():
    print(key, value)
```

##### 4.2 Dict 增

update()
一次增加多个：

```python
student = {
    "name":"Alice"
}

student.update({
    "height": 183,
    "major": "Computer Science"
})

print(student)
student.get("height")
```

### 6. Set 集合

Set（集合）的核心特点：  
无重复元素  
没有 index  
元素无固定顺序  
Mutable  

创建：  
numbers = {1, 2, 3}  

如果有重复：  
numbers = {1, 1, 2, 2, 3}  
Python 会自动去重：  
{1, 2, 3}


创建空 Set

注意：
x = {} 不是 set，而是dict  
空 set 必须：x = set()

#### 6.1 Set 查

Set 没有 index  
一般使用：print(20 in numbers)，返回True

#### 6.2 Set 增

add()
numbers = {1, 2, 3}  
numbers.add(-4)  
print(numbers)  # {1, 2, 3, -4}

update()
一次添加多个：
numbers = {1, 2}
numbers.update([3, 7, 5])  # {1, 2, 3, 5, 7}



```python

numbers = {1, 2, 3}  
numbers.add(-4)  
print(numbers)  



numbers = {1, 2}
numbers.update([3, 7, 5])
print(numbers)
```

#### 6.3 Set 删

remove()  
numbers.remove(2) #若元素不存在，会报KeyError

discard()  
numbers.discard(2)  #若元素不存在，不会报错

```python
numbers = {1,5,3,6,9,13}

numbers.remove(1)
numbers.discard(100)
print(numbers)
numbers.pop()
print(numbers)
```

#### 6.4 Set 改
由于没有index,只能先删除再添加  
numbers = {1, 2, 3}   
numbers.remove(2)   
numbers.add(20)

#### 6.5 Set的集合运算

a = {1, 2, 3}   
b = {3, 4, 5}

Union 并集  
两个集合的所有元素：  
print(a | b)  
或者：  
print(a.union(b))  # {1, 2, 3, 4, 5}


Intersection 交集  
两个集合共同拥有的：  
print(a & b)  
或者：  
print(a.intersection(b))  # {3}


Difference 差集  
A 有，但是 B 没有：  
print(a - b)  # {1, 2}


```python
a = {1, 2, 3}   
b = {3, 4, 5}

print(a | b)
print(a&b)
print(a-b)
```
