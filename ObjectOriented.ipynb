#### init() 初始化方法

实际使用 Class 时，我们通常希望对象创建时就拥有一些数据。

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

# 创建对象
student1 = Student("Alice", 22)
student2 = Student("Ning", 18)

print(student1.name)
```

#### Attribute 属性
存储在对象内部的数据

self.name  
self.age

就是：Instance Attributes（实例属性）

```python
class Student: 
    school = "ABC University" # class attribute 所有对象共享

    def __init__(self, name): 
        self.name = name      # instance attribute 单个对象拥有

student1 = Student("Alice") 
student2 = Student("Bob") 

print(student1.school) 
print(student2.school)      # student1 和 student2 都输出 ”ABC University“
```

#### Method 方法
定义在 Class 里的函数



```python
class Student: 

    height = 183
    weight = 70

    def __init__(self, name, age, score=None): 
        self.name = name 
        self.age = age
        self.score = score

    def introduce(self):
        print("My name is", self.name)

    def update_score(self, new_score): 
        self.score = new_score

    def is_passed(self): 
        return self.score >= 60

    # 定义一个类方法，指向类，而不是实例
    @classmethod
    def class_method(cls):
        print(f"I am {cls.height}cm tall.")

    # 静态方法，该方法没有访问类或实例特殊参数
    @staticmethod
    def add(a, b):
        result=a+b
        return result
    

student1 = Student("Ning", 23)
student1.introduce()

student1.update_score(99)
print(student1.score)

print(student1.is_passed())

# 测试类方法和静态方法
print("---"*10)
Student.class_method()
Student.add(1, 2)

# 测试给类动态赋值
print("---"*10)
student1.length = 15
print(student1.length)

def say_sth():
    print("I am a person.")
student1.say = say_sth
student1.say()
```

#### Inheritance 继承

一个 Class 可以继承另一个 Class 的属性和方法

```python
class Person: 
    def __init__(self, name): 
        self.name = name 

    def introduce(self): 
        print("My name is", self.name)


# 创建子类,该类可继承上面Person类的 name and introduce()
class Student(Person): 
    def __init__(self, name):
        super().__init__(name)

    def study(self): 
        print(self.name, "is studying")


student = Student("Alice")

student.introduce()
student.study()
```

#### Method Overriding 方法重写

子类可以重新定义父类的方法

```python
class Person: 
    def introduce(self): 
        print("I am a person") 

class Student(Person): 
    def introduce(self): 
        print("I am a student")

student = Student()
student.introduce()     # 输出I am a student
```

#### Encapsulation 封装

把数据和处理数据的方法放在一个 Class 中

```python
class BankAccount: 
    def __init__(self, balance): 
        self.balance = balance 

    def deposit(self, amount):  
        self.balance += amount

    def withdraw(self, amount):
        self.balance -= amount

# 外部代码只要使用，不需要关注内部实现
account1 = BankAccount(3000)

account1.deposit(2000)
print(account1.balance)

account1.withdraw(100)
print(account1.balance)
```

#### str() 
如果没有定义 _str()_ 但直接print(student)，通常会输出类似对象地址的信息

```python
# __str__() 是一个常见的特殊方法

class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Name is {self.name}, age is {self.age}"

student = Student("Alice", 22)
print(student)          # 输出 Name is Alice, age is 22
```

#### 为什么机器学习大量使用 Class？

机器学习中的模型不仅有函数，还需要保存大量内部数据。  
例如一个模型可能需要保存：  
模型参数  
weights  
bias  
训练状态  
超参数  
hyperparameters

```python
class SimpleModel: 
    def __init__(self, weight): 
        self.weight = weight 

    def predict(self, x): 
        return x * self.weight

# 创建模型：
model = SimpleModel(2)

# 预测：
prediction = model.predict(10) 
print(prediction)
```
