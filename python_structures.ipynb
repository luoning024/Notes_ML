### Python 三大程序结构  

Python 程序最基本的三种结构：  
  
顺序结构（Sequence）  
分支结构（Selection / Conditional）  
循环结构（Loop / Iteration）  



#### 1. 顺序结构 Sequence

默认从上到下依次执行

```python
name = "Alice" 
age = 22 
print(name) 
print(age) 
print("Program finished")
```

#### 2. 分支结构（Selection / Conditional）  

#### 2.1 比较运算符 Comparison Operators

== 等于   
!= 不等于   
.> 大于   
< 小于   
.>= 大于等于   
<= 小于等于  

#### 2.2 逻辑运算符 Logical Operators

and  
or  
not

```python
# AND
age = 5
has_ticket = True 
if age >= 18 and has_ticket: 
    print("Allowed")
else:
    print("shit!")

# OR
day = "Saturday" 
if day == "Saturday" or day == "Sunday": 
    print("Weekend")

# NOT 取反
logged_in = False 
if not logged_in: 
    print("Please login")
```

#### 2.3 in / not in
检查元素是否存在，经常用于条件判断。

```python
languages = ["Python", "Java", "C++"]  

if "JavaScript" not in languages:  
    print("JavaScript does not exist")
```

#### 2.4 Nested if


```python
age = 20 
has_ticket = True 
if age >= 18: 
    if has_ticket: 
        print("Allowed")
```

#### 2.5 三元表达式 Conditional Expression


```python
# 简单的if else可以放在一行 A if condition else B
age = 20

result = "Adults" if age >= 18 else "Kid"
print(result)
```

#### 2.5 match ... case

```python
day = 2 
match day: 
    case 1: 
        print("Monday") 
    case 2: 
        print("Tuesday") 
    case 3: 
        print("Wednesday") 
    case _:     # for the rest cases
        print("Unknown")
```

#### 4. 递归结构

```python
# 使用递归计算n的阶乘
# n! = n * (n-1) * ... * 1
# def fun(n):
#     if n==1:
#         return 1
#     else:
#         return n*fun(n-1)

# fun(5)  # 120

# 递归求1000以内的水仙花数
# 153 = 1^3 + 5^3 + 3^3

def shui():
    for i in range(100, 1001):
        addition = (i//100)**3 + ((i % 100) // 10)**3 + (i%10)**3
        if i == addition:
            print(f"The number is: {i}")

shui()


```
